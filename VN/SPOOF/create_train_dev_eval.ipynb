```python
import os
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
base_dir = "5000_final_virtual"
merged_dir = "merge"
os.makedirs(merged_dir, exist_ok=True)

file_count = 1
current_subdir = None  # Biến lưu trữ thư mục con hiện tại
files_copied_from_subdir = 0 # Biến đếm số file đã copy từ thư mục con

for subdir, _, files in os.walk(base_dir):
    if subdir != current_subdir:  # Kiểm tra xem có phải thư mục con mới không
        if current_subdir is not None:  # Nếu không phải thư mục đầu tiên
            print(f"Đã sao chép {files_copied_from_subdir} file từ {current_subdir}")
        current_subdir = subdir
        files_copied_from_subdir = 0 # Reset biến đếm cho thư mục con mới
        print(f"Bắt đầu duyệt thư mục: {subdir}")

    for file in files:
        if file.endswith(".flac"):
            new_filename = f"{file_count:05d}.flac"
            old_path = os.path.join(subdir, file)
            new_path = os.path.join(merged_dir, new_filename)
            shutil.copy(old_path, new_path)
            file_count += 1
            files_copied_from_subdir += 1

# In thông tin cho thư mục con cuối cùng
print(f"Đã sao chép {files_copied_from_subdir} file từ {current_subdir}")

print("Hoàn thành gộp dữ liệu.")
```

```python
# Lấy cột client_id và lọc trùng lặp
unique_client_ids = meta['client_id'].drop_duplicates().reset_index(drop=True)

# Gán lại tên cho từng ID theo định dạng LA_00001
unique_client_ids = unique_client_ids.reset_index()
unique_client_ids['new_id'] = unique_client_ids['index'].apply(lambda x: f"LA_{x+1:05d}")
unique_client_ids = unique_client_ids[['client_id', 'new_id']]

# Hiển thị kết quả
unique_client_ids
```

```python
# Tạo bản sao của meta
meta_copy = meta.copy()

# Thay thế client_id bằng new_id
meta_copy = meta_copy.merge(unique_client_ids, on='client_id', how='left')
meta_copy['client_id'] = meta_copy['new_id']
meta_copy = meta_copy.drop(columns=['new_id'])
```

```python
# Nhân bản dữ liệu 11 lần (10 lần spoof + 1 lần bonafide)
meta_expanded = pd.concat([meta_copy] * 11, ignore_index=True)

# Đổi tên file và thêm cột code
codes = ['bonafide', 'A01', 'A02', 'A03', 'A05', 'A06', 'A08', 'A10', 'A12', 'A15', 'A19']
meta_expanded['code'] = [codes[i // 5000] for i in range(len(meta_expanded))]
meta_expanded['path'] = meta_expanded.apply(
    lambda row: f"{int(row.name % 5000 + 1 + (row.name // 5000) * 5000):05d}.flac", axis=1
)
```

```python
# Thử nghiệm mới
import os
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split

base_dir = "/home/coronatusvi/Documents/GITHUB/VoiceAntiSpoofingAI/VN/SPOOF/merge"
output_dir = "/home/coronatusvi/Documents/GITHUB/VoiceAntiSpoofingAI/VN/SPOOF/LA"

# Đọc metadata
metadata_path = os.path.join(base_dir, "/home/coronatusvi/Documents/GITHUB/VoiceAntiSpoofingAI/VN/SPOOF/meta_expanded.tsv")
metadata = pd.read_csv(metadata_path, sep='\t')

# Chia dữ liệu thành train, dev, eval
train_files, temp_files = train_test_split(metadata, test_size=0.3, random_state=42)
dev_files, eval_files = train_test_split(temp_files, test_size=0.5, random_state=42)

# Tạo thư mục đầu ra
os.makedirs(output_dir, exist_ok=True)
os.makedirs(os.path.join(output_dir, "ASVspoof2019_LA_cm_protocols"), exist_ok=True)
os.makedirs(os.path.join(output_dir, "ASVspoof2019_LA_asv_protocols"), exist_ok=True)
os.makedirs(os.path.join(output_dir, "ASVspoof2019_LA_asv_scores"), exist_ok=True) # Thêm thư mục scores
os.makedirs(os.path.join(output_dir, "ASVspoof2019_LA_train", "flac"), exist_ok=True) # Thêm thư mục flac trong train
os.makedirs(os.path.join(output_dir, "ASVspoof2019_LA_dev", "flac"), exist_ok=True) # Thêm thư mục flac trong dev
os.makedirs(os.path.join(output_dir, "ASVspoof2019_LA_eval", "flac"), exist_ok=True) # Thêm thư mục flac trong eval

# Tạo metadata cho thư mục LA_cm_protocols
def copy_and_create_LA_cm_protocol(data, output_folder, protocol_file, data_id, gender=None):
    """
    Sao chép tệp âm thanh vào thư mục flac, đổi tên tệp và tạo tệp giao thức (protocol).
    
    Args:
        data (DataFrame): Dữ liệu metadata.
        output_folder (str): Tên thư mục đầu ra (train, dev, eval).
        protocol_file (str): Đường dẫn tệp giao thức cần tạo.
        data_id (str): Tiền tố định danh dữ liệu (VD: LA_T, LA_D, LA_E).
        gender (str, optional): Lọc dữ liệu theo giới tính (male, female). Mặc định là None.
    """
    if gender:
        data = data[data['gender'] == gender]

    # Đường dẫn thư mục flac
    flac_destination = os.path.join(output_dir, f"ASVspoof2019_LA_{output_folder}", "flac")
    os.makedirs(flac_destination, exist_ok=True)

    with open(protocol_file, 'w') as f_protocol:
        for index, row in data.iterrows():
            file_path = os.path.join(base_dir, row['path'])
            if os.path.exists(file_path):
                # Đổi tên tệp theo định dạng LA_XXXX.flac
                client_id = row['client_id']
                new_file_name = f"{data_id}_{index + 1:04d}"
                new_file_path = os.path.join(flac_destination, f"{new_file_name}.flac")

                # Sao chép tệp và đổi tên
                shutil.copy(file_path, new_file_path)

                # Xác định nhãn (bonafide hoặc spoof)
                label = row['code'] if row['code'] != "bonafide" else "-" # Lấy nhãn từ cột 'code'
                target_type = "bonafide" if row['code'] == "bonafide" else "spoof"

                # Ghi thông tin vào tệp giao thức
                f_protocol.write(f"{client_id} {data_id}_{index + 1:04d} - {label} {target_type}\n")

# Tạo tệp giao thức và sao chép dữ liệu
copy_and_create_LA_cm_protocol(train_files, "train", os.path.join(output_dir, "ASVspoof2019_LA_cm_protocols", "ASVspoof2019.LA.cm.train.trn.txt"), "LA_T")
copy_and_create_LA_cm_protocol(dev_files, "dev", os.path.join(output_dir, "ASVspoof2019_LA_cm_protocols", "ASVspoof2019.LA.cm.dev.trl.txt"), "LA_D")
copy_and_create_LA_cm_protocol(eval_files, "eval", os.path.join(output_dir, "ASVspoof2019_LA_cm_protocols", "ASVspoof2019.LA.cm.eval.trl.txt"), "LA_E")

# Tạo metadata cho thư mục LA_asv_protocols
def create_LA_asv_protocol(data, protocol_file, data_id, gender=None):
    """
    Tạo tệp giao thức ASV với thông tin từ metadata.
    
    Args:
        data (DataFrame): Dữ liệu metadata.
        protocol_file (str): Đường dẫn tệp giao thức cần tạo.
        data_id (str): Tiền tố định danh dữ liệu (VD: LA_T, LA_D, LA_E).
        gender (str, optional): Lọc dữ liệu theo giới tính (male, female). Mặc định là None.
    """
    if gender:
        data = data[data['gender'] == gender]

    with open(protocol_file, 'w') as f_protocol:
        for index, row in data.iterrows():
            # Lấy client_id từ metadata
            client_id = row['client_id']

            # Đổi tên tệp theo định dạng LA_XXXX.flac
            new_file_name = f"{data_id}_{index + 1:04d}"

            # Lấy nhãn (code) từ metadata
            label = row['code']

            # Xác định target_type
            if label == "bonafide":
                target_type = "target"  # Hoặc "nontarget" tùy thuộc vào ngữ cảnh
            else:
                target_type = "spoof"

            # Ghi thông tin vào tệp giao thức
            f_protocol.write(f"{client_id} {new_file_name} {label} {target_type}\n")

# Tạo tệp giao thức ASV (dev, eval)
create_LA_asv_protocol(dev_files, os.path.join(output_dir, "ASVspoof2019_LA_asv_protocols", "ASVspoof2019.LA.asv.dev.female.trl.txt"), 'LA_D', gender='female')
create_LA_asv_protocol(dev_files, os.path.join(output_dir, "ASVspoof2019_LA_asv_protocols", "ASVspoof2019.LA.asv.dev.male.trl.txt"), 'LA_D', gender='male')
create_LA_asv_protocol(eval_files, os.path.join(output_dir, "ASVspoof2019_LA_asv_protocols", "ASVspoof2019.LA.asv.eval.male.trl.txt"), 'LA_E', gender='male')
create_LA_asv_protocol(eval_files, os.path.join(output_dir, "ASVspoof2019_LA_asv_protocols", "ASVspoof2019.LA.asv.eval.female.trl.txt"), 'LA_E', gender='female')
```

```python
# FEMALE DEV
protocol_file_path = "LA/ASVspoof2019_LA_asv_protocols/ASVspoof2019.LA.asv.dev.female.trl.txt"

# Read the file into a pandas DataFrame
protocol_data = pd.read_csv(protocol_file_path, sep=" ", header=None, names=["client_id", "file_name", "label", "target_type"])

# Nhóm các file theo client_id
grouped_files = protocol_data.groupby('client_id')['file_name'].apply(lambda x: ','.join(x)).reset_index()

# Định dạng lại dữ liệu
formatted_data = grouped_files.apply(lambda row: f"{row['client_id']} {row['file_name']}", axis=1)

# Định dạng lại dữ liệu và ghi vào file FEMALE DEV
output_file_path = "LA/ASVspoof2019_LA_asv_protocols/ASVspoof2019.LA.asv.dev.female.trn.txt"

# Ghi kết quả vào file
with open(output_file_path, 'w') as f:
    for line in formatted_data:
        f.write(line + '\n')

print(f"Kết quả đã được ghi vào file: {output_file_path}")
```
