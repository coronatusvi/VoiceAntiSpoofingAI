# A08
✅ High-pass filter (lọc bỏ tần số thấp nhẹ)
✅ Formant shifting (thay đổi vị trí formant, giúp giọng khác đi một chút)

```python
import librosa
import librosa.display
import numpy as np
import soundfile as sf
import scipy.signal
```

```python
# Load audio
y, sr = librosa.load("5000_final_virtual/clip_flac_spoof_A01_5000/common_voice_vi_21833313.flac", sr=16000)

# 1️⃣ High-pass filter (Áp dụng bộ lọc thông cao)
b, a = scipy.signal.butter(N=4, Wn=100, btype='high', fs=sr)
y_filtered = scipy.signal.filtfilt(b, a, y)

# 2️⃣ Thay đổi formant (Shift formant bằng cách thay đổi hệ số pre-emphasis)
y_formant_shifted = librosa.effects.preemphasis(y_filtered, coef=0.95)

# Lưu lại thành A08
sf.write("A08.flac", y_formant_shifted, sr)
```

# A10

- Pre-emphasis Filtering (Lọc nhấn mạnh tần số cao)
- Time-stretching (Thay đổi tốc độ mà không đổi cao độ)
- Pitch Shifting (Thay đổi cao độ giọng nói)

```python
input_path = "5000_final_virtual/clip_flac_spoof_A01_5000/common_voice_vi_21964747.flac"

# Load audio
y, sr = librosa.load(input_path, sr=16000)

# 1️⃣ Pre-emphasis Filtering (Lọc nhấn mạnh tần số cao)
y_preemphasized = librosa.effects.preemphasis(y, coef=0.1)

# 2️⃣ Time-stretching (Tiến trình kéo dài thời gian)
y_stretched = librosa.effects.time_stretch(y_preemphasized, rate=1.01)

# 3️⃣ Pitch Shifting (Tăng cao độ lên 2 bán âm)
y_pitch_shifted = librosa.effects.pitch_shift(y_stretched, sr=sr, n_steps=2)

# Lưu lại thành A10
sf.write("A10.flac", y_pitch_shifted, sr)
```

```python
# Đường dẫn thư mục gốc và thư mục đích
input_dir = "5000_final_virtual/clip_flac_spoof_A01_5000"
output_dir = "5000_final_virtual/clip_flac_spoof_A10_5000"

# Tạo thư mục đích nếu chưa tồn tại
os.makedirs(output_dir, exist_ok=True)

# Lặp qua tất cả các file trong thư mục đầu vào
for filename in os.listdir(input_dir):
    if filename.endswith(".flac"):  # Chỉ xử lý các file .flac
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)  # Lưu với cùng tên file

        # Load audio
        y, sr = librosa.load(input_path, sr=16000)

        # 1️⃣ Pre-emphasis Filtering (Lọc nhấn mạnh tần số cao)
        y_preemphasized = librosa.effects.preemphasis(y, coef=0.2)

        # 2️⃣ Time-stretching (Tăng tốc độ nói 5% mà không đổi cao độ)
        y_stretched = librosa.effects.time_stretch(y_preemphasized, rate=1.01)

        # 3️⃣ Pitch Shifting (Tăng cao độ lên 2 bán âm)
        y_pitch_shifted = librosa.effects.pitch_shift(y_stretched, sr=sr, n_steps=2)

        # Lưu lại file đã xử lý vào thư mục đích
        sf.write(output_path, y_pitch_shifted, sr)

print("🎉 Hoàn thành xử lý tất cả các file sang A10")
```

# A12

- Additive Noise Augmentation (Thêm nhiễu Gaussian)	Làm méo nhẹ tín hiệu bằng cách thêm nhiễu	y + np.random.normal(0, 0.005, y.shape)
- Dynamic Range Compression (DRC) (Nén dải động)	Cân bằng mức âm lượng giữa các phần khác nhau của audio	np.interp(y, (y.min(), y.max()), (-1, 1))
- Reverberation Effect (Hiệu ứng vang)	Thêm âm vang để làm giả không gian thu âm	scipy.signal.convolve(y, impulse_response, mode="full")
- Spectral Subtraction (Khử nhiễu bằng trừ phổ)

```python
import scipy.signal
```

```python
input_path = "5000_final_virtual/clip_flac_spoof_A01_5000/common_voice_vi_21833331.flac"

# Load audio
y, sr = librosa.load(input_path, sr=16000)

# 1️⃣ Additive Noise Augmentation (Thêm nhiễu Gaussian) - Nhẹ hơn
noise = np.random.normal(0, 0.0008, y.shape)  # Giảm nhiễu để tránh mất tiếng
y_noisy = y + noise

# 2️⃣ Dynamic Range Compression - Giữ nguyên biên độ tốt hơn
y_compressed = np.interp(y_noisy, (y_noisy.min(), y_noisy.max()), (-0.98, 0.98))

# 3️⃣ Reverberation Effect - Giảm ảnh hưởng
impulse_response = np.exp(-np.linspace(0, 1, 80))  # Giảm chiều dài impulse response
y_reverbed = scipy.signal.fftconvolve(y_compressed, impulse_response, mode="same")  

# 4️⃣ **(TẮT Spectral Subtraction)**
y_final = y_reverbed / np.max(np.abs(y_reverbed)) * 0.98  # Chuẩn hóa nhẹ hơn

# Lưu lại thành A12
sf.write("A12.flac", y_final, sr)
```

```python
# Đường dẫn thư mục gốc và thư mục đích
input_dir = "5000_final_virtual/clip_flac_spoof_A01_5000"
output_dir = "5000_final_virtual/clip_flac_spoof_A12_5000"

# Tạo thư mục đích nếu chưa tồn tại
os.makedirs(output_dir, exist_ok=True)

# Lặp qua tất cả các file trong thư mục đầu vào
for filename in os.listdir(input_dir):
    if filename.endswith(".flac"):  # Chỉ xử lý các file .flac
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)  # Lưu với cùng tên file

        # Load audio
        y, sr = librosa.load(input_path, sr=16000)

        # 1️⃣ Additive Noise Augmentation (Thêm nhiễu Gaussian) - Nhẹ hơn
        noise = np.random.normal(0, 0.0008, y.shape)  # Giảm nhiễu để tránh mất tiếng
        y_noisy = y + noise

        # 2️⃣ Dynamic Range Compression - Giữ nguyên biên độ tốt hơn
        y_compressed = np.interp(y_noisy, (y_noisy.min(), y_noisy.max()), (-0.98, 0.98))

        # 3️⃣ Reverberation Effect - Giảm ảnh hưởng
        impulse_response = np.exp(-np.linspace(0, 1, 80))  # Giảm chiều dài impulse response
        y_reverbed = scipy.signal.fftconvolve(y_compressed, impulse_response, mode="same")  

        # 4️⃣ **(TẮT Spectral Subtraction)**
        y_final = y_reverbed / np.max(np.abs(y_reverbed)) * 0.98  # Chuẩn hóa nhẹ hơn

        # Lưu lại file đã xử lý vào thư mục đích
        sf.write(output_path, y_final, sr)

print("🎉 Hoàn thành xử lý tất cả các file sang A12")
```
